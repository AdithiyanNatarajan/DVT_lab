```python
import pandas as pd

# Read the uploaded CSV file
df = pd.read_csv(r"D:\fifa\2024\players.csv")

# Print the first 5 rows
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_overall = df[df["Overall"] > 90]
print(high_overall[["Name", "Club", "Position", "Overall"]])
```

```python
strikers = df[df["Position"] == "ST"]
print(strikers[["Name", "Club", "Position", "Overall"]].head())
```

```python
sorted_df = df.sort_values(by="Overall", ascending=False)

# Display top 5 results with key columns
print(sorted_df[["Name", "Club", "Position", "Age", "Overall"]].head())
```

```python
import matplotlib.pyplot as plt

sorted_df = df.sort_values(by="Overall", ascending=False)
top10 = sorted_df.head(10)

plt.figure(figsize=(10, 5))
plt.bar(top10["Name"], top10["Overall"], color='skyblue', edgecolor='black')
plt.title("Top 10 Players by Overall Rating")
plt.xlabel("Player Name")
plt.ylabel("Overall Rating")
plt.ylim(85, 95)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
