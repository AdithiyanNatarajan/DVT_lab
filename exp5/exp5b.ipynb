```python
import pandas as pd
import matplotlib.pyplot as plt

# Group by Age and calculate the sum of Overall ratings
df = pd.read_csv(r"D:\fifa\2024\players.csv")
age_group = df.groupby("Age")["Overall"].sum()

# Plot line chart
plt.figure(figsize=(10, 5))
plt.plot(age_group.index, age_group.values, marker="o", color="teal")
plt.title("Total Overall Rating by Age")
plt.xlabel("Age")
plt.ylabel("Total Overall Rating")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count players by preferred foot
foot_counts = df["Preferred foot"].value_counts()

# Plot bar chart
plt.figure(figsize=(8, 5))
plt.bar(
    foot_counts.index,
    foot_counts.values,
    color=["navy", "orange"],
    edgecolor="black",
)
plt.title("Player Count by Preferred Foot")
plt.xlabel("Preferred Foot")
plt.ylabel("Number of Players")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count values for Attacking Work Rate
work_rate_counts = df["Att work rate"].value_counts()

# Plot pie chart
plt.figure(figsize=(7, 7))
plt.pie(
    work_rate_counts,
    labels=work_rate_counts.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=["#ff9999", "#66b3ff", "#99ff99"],
)
plt.title("Attacking Work Rate Distribution")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Select numerical column
pace = df["Pace"].dropna()

# Plot histogram with 5 bins
plt.figure(figsize=(10, 5))
plt.hist(pace, bins=5, color="coral", edgecolor="black", alpha=0.7)
plt.title("Distribution of Player Pace Ratings")
plt.xlabel("Pace Rating")
plt.ylabel("Frequency")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
