```python
import pandas as pd
import matplotlib.pyplot as plt

# Group by Age and calculate the average Overall rating
df = pd.read_csv(r"D:\fifa\2024\players.csv")
age_overall = df.groupby("Age")["Overall"].mean().reset_index()


# Plot line chart
plt.figure(figsize=(10, 5))
plt.plot(
    age_overall["Age"],
    age_overall["Overall"],
    marker="o",
    color="b",
    linestyle="-",
)
plt.title("Average Overall Rating by Age")
plt.xlabel("Age")
plt.ylabel("Average Overall Rating")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Count players by position
position_counts = df["Position"].value_counts().head(6)

# Plot bar chart
plt.figure(figsize=(10, 5))
plt.bar(
    position_counts.index,
    position_counts.values,
    color="teal",
    edgecolor="black",
)
plt.title("Player Count by Top Positions")
plt.xlabel("Position")
plt.ylabel("Number of Players")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
departments = ["HR", "IT", "Finance", "Sales"]
employees = [25, 40, 20, 15]
plt.pie(employees,labels=departments,autopct="%1.1f%%")
plt.title("Employee Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Histogram of player overall ratings
plt.figure(figsize=(10, 5))
plt.hist(df["Overall"], bins=15, color="purple", edgecolor="black", alpha=0.7)
plt.title("Distribution of Player Overall Ratings")
plt.xlabel("Overall Rating")
plt.ylabel("Frequency")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
