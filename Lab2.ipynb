```python
!pip install openpyxl
import pandas as pd
import sqlite3

# Import data from CSV
df = pd.read_csv(r"D:\fifa\2024\all_players.csv")

print("First 5 Rows")
print(df.head())

print("\nLast 5 Rows")
print(df.tail())

print("\nDataset Information")
print(df.info())

print("\nSummary Statistics")
print(df.describe())

print("\nShape of Dataset")
print(df.shape)

print("\nColumn Names")
print(df.columns)

print("\nMissing Values")
print(df.isnull().sum())

# Convert CSV to Excel
df.to_excel("all_players.xlsx", index=False)

# Import data from Excel
excel_df = pd.read_excel("all_players.xlsx")

print("\nData Imported from Excel")
print(excel_df.head())

# Create SQLite Database
conn = sqlite3.connect("players.db")

# Store data in SQL database
df.to_sql("players", conn, if_exists="replace", index=False)

# Import data from SQL
sql_df = pd.read_sql("SELECT * FROM players", conn)

print("\nData Imported from SQL")
print(sql_df.head())

conn.close()

# Import data from Web (Optional)
# Replace the URL with your own dataset URL
"""
url = "https://raw.githubusercontent.com/username/repository/main/all_players.csv"
web_df = pd.read_csv(url)
print("\nData Imported from Web")
print(web_df.head())
"""

# Export data to CSV
df.to_csv("players_output.csv", index=False)

# Export data to Excel
df.to_excel("players_output.xlsx", index=False)

# Export data to JSON
df.to_json("players_output.json", orient="records")

print("\nData exported successfully to CSV, Excel, and JSON.")
```
