```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv(r"D:\fifa\2024\players.csv")

# Display dataset
print(df)
```

```python
print(df.head())

# Display shape
print(df.shape)

# Display data types
print(df.dtypes)
```

```python
print(df.describe())

# Filter students with marks greater than 80
print(df[df["Pace"] > 70])

# Filter students with attendance greater than 90
print(df[df["Age"] > 30])
```

```python
print(df.sort_values("Shooting", ascending=False))

# Sort students by attendance
print(df.sort_values("Passing", ascending=False))

# Department-wise average marks
print(df.groupby("Nation")["Shooting"].mean())
```

```python


# Calculate mean
print("Mean Overall:", df["Overall"].mean())

# Calculate median
print("Median Overall:", df["Overall"].median())

# Calculate minimum
print("Minimum Overall:", df["Overall"].min())

# Calculate maximum
print("Maximum Overall:", df["Overall"].max())

# Calculate standard deviation
print("Standard Deviation:", df["Overall"].std())

# Display correlation matrix for key player stats
print(df[["Overall", "Pace", "Shooting", "Passing", "Age"]].corr())

# Display correlation between Overall and Pace
print("Correlation:", df["Overall"].corr(df["Pace"]))
```

```python
top10 = df.nlargest(10, "Overall")

# Generate Bar Chart
plt.figure(figsize=(10, 6))
plt.bar(top10["Name"], top10["Overall"], color="skyblue", edgecolor="black")
plt.title("Top 10 Players by Overall Rating", fontsize=14, fontweight="bold")
plt.xlabel("Player Name", fontsize=12)
plt.ylabel("Overall Rating", fontsize=12)
plt.ylim(85, 95)  # Zoom in to emphasize rating differences
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
top10 = df.nlargest(10, "Overall")

# Generate Line Plot
plt.figure(figsize=(10, 6))
plt.plot(
    top10["Name"],
    top10["Pace"],
    marker="o",
    color="crimson",
    linewidth=2,
    markersize=8,
)
plt.title("Pace Rating of Top 10 Overall Players", fontsize=14, fontweight="bold")
plt.xlabel("Player Name", fontsize=12)
plt.ylabel("Pace Rating", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(
    df["Pace"], df["Overall"], alpha=0.3, color="dodgerblue", edgecolors="none", s=15
)
plt.title("Overall Rating vs. Pace", fontsize=14, fontweight="bold")
plt.xlabel("Pace", fontsize=12)
plt.ylabel("Overall Rating", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))
n, bins, patches = plt.hist(
    df["Overall"], bins=5, color="teal", edgecolor="black", alpha=0.7
)
plt.title("Overall Rating Distribution", fontsize=14, fontweight="bold")
plt.xlabel("Overall Rating", fontsize=12)
plt.ylabel("Number of Players", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Annotate counts on top of bars
for i in range(len(n)):
    plt.text(
        (bins[i] + bins[i + 1]) / 2,
        n[i] + 100,
        str(int(n[i])),
        ha="center",
        va="bottom",
    )

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 6))
plt.boxplot(df["Overall"], patch_artist=True, boxprops=dict(facecolor="lightskyblue"))
plt.title("Overall Rating Box Plot", fontsize=14, fontweight="bold")
plt.ylabel("Overall Rating", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()

# Option 2: Multi-Attribute Comparison Box Plot
stats = [
    "Overall",
    "Pace",
    "Shooting",
    "Passing",
    "Dribbling",
    "Defending",
    "Physicality",
]

plt.figure(figsize=(10, 6))
plt.boxplot(
    [df[col] for col in stats],
    labels=stats,
    patch_artist=True,
    boxprops=dict(facecolor="lightskyblue", color="navy"),
    medianprops=dict(color="crimson", linewidth=2),
    flierprops=dict(marker="o", markersize=3, alpha=0.3),
)
plt.title("Distribution of Key Player Ratings", fontsize=14, fontweight="bold")
plt.ylabel("Rating Value", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
